```python
#!/usr/bin/env python3
import os
import re
import time
import textwrap
import csv
from io import StringIO
from urllib.parse import urlparse

import feedparser
from dateutil import parser as dtparse
from pytz import timezone as pytz_timezone
from openai import OpenAI

# -----------------------
# Config
# -----------------------

MAX_ITEMS = 200
OUTPUT_PSV = "./asx.psv"

# Pipe separated values, 7 columns
DELIM = "|"
HEADER = "date|time|headline|newspaper|sector|stock|article link"
EXPECTED_COLS = 7

TZ = pytz_timezone(os.getenv("LOCAL_TZ", "Australia/Sydney"))
RETRY_LIMIT = int(os.getenv("RETRY_LIMIT", "3"))
VALIDATION_RETRIES = int(os.getenv("VALIDATION_RETRIES", "2"))

def get_env_str(name: str, default: str) -> str:
    v = os.getenv(name)
    return v.strip() if v else default

MODEL = get_env_str("OPENAI_MODEL", "gpt-5")

# -----------------------
# Helpers
# -----------------------
def backoff_retry(func, *args, **kwargs):
    for attempt in range(1, RETRY_LIMIT + 1):
        try:
            return func(*args, **kwargs)
        except Exception as e:
            if attempt == RETRY_LIMIT:
                print(f"ERROR: {func.__name__} failed after {RETRY_LIMIT} attempts: {e}")
                return None
            wait = 5 * (3 ** (attempt - 1))
            print(f"Warning: {func.__name__} failed ({e}), retrying in {wait}s...")
            time.sleep(wait)

def parse_time(entry):
    for attr in ("published", "updated"):
        val = getattr(entry, attr, None)
        if val:
            try:
                dt = dtparse.parse(val).astimezone(TZ)
                date = dt.strftime("%Y-%m-%d")
                minute = dt.strftime("%H:%M")
                return date, minute
            except Exception:
                pass
    return "", ""

def clean_link(link):
    return re.sub(r"[?&].*$", "", (link or "").strip())

def _host_from_url(url: str) -> str:
    try:
        host = urlparse(url).netloc.lower()
        if host.startswith("www."):
            host = host[4:]
        return host
    except Exception:
        return ""

def collect_feed_items():
    items = []
    for url in RSS_FEEDS:
        feed = backoff_retry(feedparser.parse, url)
        if not feed:
            continue
        # newspaper name: prefer feed title, else host of feed URL
        feed_title = getattr(feed.feed, "title", "").strip() if getattr(feed, "feed", None) else ""
        fallback_newspaper = feed_title or _host_from_url(url)
        for e in getattr(feed, "entries", []):
            date, minute = parse_time(e)
            link = clean_link(getattr(e, "link", ""))
            # Prefer entry-level source title if present, else feed title fallback
            entry_source = ""
            try:
                # Some feeds include e.source.title
                src = getattr(e, "source", None)
                if src and getattr(src, "title", None):
                    entry_source = src.title.strip()
            except Exception:
                pass
            newspaper = entry_source or feed_title or _host_from_url(link) or fallback_newspaper

            items.append({
                "date": date,
                "time": minute,
                "headline": getattr(e, "title", "").strip(),
                "newspaper": newspaper,
                "link": link,
            })
    # dedupe by link and sort newest first
    seen, out = set(), []
    for it in items:
        if not it["link"] or it["link"] in seen:
            continue
        seen.add(it["link"])
        out.append(it)
    out.sort(key=lambda x: (x["date"], x["time"]), reverse=True)
    return out[:MAX_ITEMS]

# -----------------------
# Prompting
# -----------------------
SYSTEM_PROMPT = (
    "You map Australian finance headlines to one ASX ticker and sector.\n"
    "Return PSV only with 7 columns matching the provided header.\n"
    "If unknown, leave sector or stock blank."
)

def build_user_prompt(items):
    parts = [
        "Produce PSV (pipe-separated values) for these articles in the SAME order.",
        "Fields per article: date, time, headline, newspaper, sector, stock, article link.",
        "Return PSV ONLY starting with the EXACT header:",
        HEADER,
        "",
    ]
    for i, it in enumerate(items, 1):
        parts.append(f"## ARTICLE {i}")
        parts.append(f"date: {it['date']}")
        parts.append(f"time: {it['time']}")
        parts.append(f"headline: {it['headline']}")
        parts.append(f"newspaper: {it['newspaper']}")
        parts.append("sector: ")   # leave empty if unknown
        parts.append("stock: ")    # leave empty if unknown
        parts.append(f"article link: {it['link']}")
        parts.append("")
    example = textwrap.dedent(f"""\
        Example of exactly one row (illustrative only):
        {HEADER}
        2025-10-21|09:00|Sample headline|Sydney Morning Herald|||https://example.com/article
    """)
    parts.append(example)
    parts.append(
        "Rules:\n"
        "1) PSV only. No prose. No code fences.\n"
        f"2) First line must be exactly: {HEADER}\n"
        "3) Exactly 7 columns per row. Do not repeat the header.\n"
        "4) Quote fields that contain a pipe or newline using double quotes. Escape inner quotes by doubling them.\n"
        "5) If sector or stock is unknown, leave it empty."
    )
    return "\n".join(parts)

def call_model(prompt_text):
    client = OpenAI()
    return backoff_retry(lambda: client.responses.create(
        model=MODEL,
        input=[
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": prompt_text},
        ],
    ).output_text)

# -----------------------
# PSV validation and cleanup
# -----------------------
def _strip_empty_lines_keep_order(text: str) -> str:
    return "\n".join([ln for ln in text.splitlines() if ln.strip()])

def sanitize_model_psv(psv_text: str, header: str) -> str:
    if not psv_text:
        return header + "\n"
    lines = [ln for ln in psv_text.splitlines() if ln.strip()]
    # trim to first exact header
    try:
        idx = next(i for i, ln in enumerate(lines) if ln.strip() == header)
        text = "\n".join(lines[idx:]) + "\n"
    except StopIteration:
        text = "\n".join(lines) + "\n"
    # keep only first header
    out, seen = [], False
    for ln in text.splitlines():
        if ln.strip() == header:
            if seen:
                continue
            seen = True
        out.append(ln)
    return "\n".join(out) + "\n"

def validate_psv(psv_text: str, header: str, expected_cols: int) -> tuple[bool, str]:
    if not psv_text or not psv_text.strip():
        return False, "Empty response"
    text = _strip_empty_lines_keep_order(psv_text)
    lines = text.splitlines()
    if not lines:
        return False, "No lines"
    if lines[0].strip() != header:
        return False, "Header mismatch"
    reader = csv.reader(StringIO("\n".join(lines)), delimiter=DELIM, quotechar='"')
    hdr = next(reader, None)
    if hdr is None:
        return False, "No rows"
    if len(hdr) != expected_cols:
        return False, f"Header has {len(hdr)} columns, expected {expected_cols}"
    line_no = 1
    for row in reader:
        line_no += 1
        if not any((cell or "").strip() for cell in row):
            continue
        if len(row) != expected_cols:
            return False, f"Row {line_no} has {len(row)} columns, expected {expected_cols}"
    return True, ""

# -----------------------
# Save
# -----------------------
def save_psv_validated(psv_text: str, path: str):
    cleaned = sanitize_model_psv(psv_text, HEADER)
    ok, reason = validate_psv(cleaned, HEADER, EXPECTED_COLS)
    if not ok:
        print(f"Final validation failed: {reason}. Writing header only.")
        cleaned = HEADER + "\n"
    # normalise UNMAPPED token to blank in sector or stock
    cleaned = re.sub(r'\|UNMAPPED\|', '||', cleaned)
    with open(path, "w", encoding="utf-8", newline="") as f:
        f.write(cleaned)
    rows = max(0, len([ln for ln in cleaned.splitlines()[1:] if ln.strip()]))
    print(f"Wrote {rows} rows to {path}")

# -----------------------
# Main
# -----------------------
def main():
    items = collect_feed_items()
    if not items:
        print("No feed items found.")
        save_psv_validated(HEADER + "\n", OUTPUT_PSV)
        return

    prompt = build_user_prompt(items)
    psv_text = call_model(prompt)

    for _ in range(VALIDATION_RETRIES):
        psv_text = sanitize_model_psv(psv_text or "", HEADER)
        ok, reason = validate_psv(psv_text, HEADER, EXPECTED_COLS)
        if ok:
            break
        retry_prompt = (
            prompt
            + "\n\nCRITICAL OUTPUT RULES:\n"
            + "PSV only. No prose. No code fences.\n"
            + f"First line must be exactly: {HEADER}\n"
            + "Exactly 7 columns per row. Do not repeat the header.\n"
            + "Quote fields with '|' or newlines using double quotes.\n"
            + f"\nYour last output failed validation: {reason}\n"
            + "Re-output the PSV now, corrected."
        )
        psv_text = call_model(retry_prompt)

    save_psv_validated(psv_text or "", OUTPUT_PSV)

if __name__ == "__main__":
    main()
```
